```python
import gradio as gr
import pandas as pd
import joblib


model = joblib.load("fraud_detection_pipeline.pkl")


def predict_fraud(transaction_type, amount, oldbalanceOrg, newbalanceOrig, oldbalanceDest, newbalanceDest):

    input_data = pd.DataFrame([{
        "type": transaction_type,
        "amount": amount,
        "oldbalanceOrg": oldbalanceOrg,
        "newbalanceOrig": newbalanceOrig,
        "oldbalanceDest": oldbalanceDest,
        "newbalanceDest": newbalanceDest
    }])


    prediction = model.predict(input_data)[0]


    if prediction == 1:
        return "⚠️ This transaction may be FRAUDULENT."
    else:
        return "✅ This transaction looks LEGITIMATE."


iface = gr.Interface(
    fn=predict_fraud,
    inputs=[
        gr.Dropdown(["PAYMENT", "TRANSFER", "CASH_OUT", "DEPOSIT"], label="Transaction Type"),
        gr.Number(label="Amount", value=1000.0),
        gr.Number(label="Old Balance (Sender)", value=10000.0),
        gr.Number(label="New Balance (Sender)", value=9000.0),
        gr.Number(label="Old Balance (Receiver)", value=0.0),
        gr.Number(label="New Balance (Receiver)", value=0.0),
    ],
    outputs=gr.Textbox(label="Prediction Result"),
    title="💳 Fraud Detection Prediction App",
    description="Enter transaction details below and click **Submit** to predict whether it might be fraudulent.",
    theme="default"
)


iface.launch()
```
